```python
# 1.Create an array that is having user defined inputs and with the help of for loop, fetch all the prime numbers and print the numbers

arr = []


n = int(input("Enter how many numbers: "))


for i in range(n):
    val = int(input(f"Enter element {i+1}: "))
    arr.append(val)

print("Prime numbers are:")


for num in arr:
    if num > 1:
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num)
```

# 1. Create a function named ‘factor’ that can only accept 1 argument. 
# unction should return the factorial of that number.


```python
def factor(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factor(n - 1)


print(factor(5)) 
```

# 2. Create a function named ‘check_string’, the function should accept a string

# the letter ‘s’ in it. If it contains the letter ‘s’ then print- ‘The string is
# containing the letter ‘s’’, if not then print- ‘The string doesn’t contain the

```python
def check_string(user_input):
    if 's' in user_input.lower():
        print("The string is containing the letter 's'")
    else:
        print("The string doesn't contain the letter 's'")


check_string("Python Script")
```

# 5. Take user input string and check whether that string is palindrome or not.

```python
user_str = input("Enter a string to check for palindrome: ")


clean_str = user_str.replace(" ", "").lower()

if clean_str == clean_str[::-1]:
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```
