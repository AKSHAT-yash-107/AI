```python
# 3 Write a python program to check whether a number is Armstrong number or not using

num= int(input("Enter a number: "))
temp =num
result=0
count=0
while(temp!=0):
    r=temp%10
    num=num//10
    count+=1
    
while(num!=0):
    r=num%10
    result+=r**count
    num/=10

if result==num:
    print("armstrong number")
else:
    print("not")
    
```

```python
# 5 Write a python program to find first ‘n’ terms of the Fibonacci series.

n = int(input("Enter number of terms: "))

a = 0
b = 1

print("Fibonacci series:")

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
# 6 Write a python program to print the reverse of a given number using loop.
num = int(input("Enter a number: "))

rev = 0
temp = abs(num)   

while temp != 0:
    r = temp % 10
    rev = rev * 10 + r
    temp //= 10

if num < 0:
    rev = -rev    

print("Reversed number:", rev)
```

```python
def decimal_to_binary(n):
    if n == 0:
        return "0"
    binary = ""
    while n > 0:
        binary = str(n % 2) + binary
        n //= 2
    return binary

def decimal_to_hex(n):
    if n == 0:
        return "0"
    digits = "0123456789ABCDEF"
    hexa = ""
    while n > 0:
        hexa = digits[n % 16] + hexa
        n //= 16
    return hexa

def hex_to_decimal(hex_value):
    hex_value = hex_value.upper()
    digits = "0123456789ABCDEF"
    decimal = 0
    for char in hex_value:
        decimal = decimal * 16 + digits.index(char)
    return decimal

def decimal_to_octal(n):
    if n == 0:
        return "0"
    octal = ""
    while n > 0:
        octal = str(n % 8) + octal
        n //= 8
    return octal

def hex_to_octal(hex_value):
    dec = hex_to_decimal(hex_value)
    return decimal_to_octal(dec)

def binary_to_decimal(binary):
    decimal = 0
    for digit in binary:
        decimal = decimal * 2 + int(digit)
    return decimal

print(decimal_to_binary(25))      
print(decimal_to_hex(254))        
print(hex_to_octal("1A3"))        
print(binary_to_decimal("1011"))  
```
